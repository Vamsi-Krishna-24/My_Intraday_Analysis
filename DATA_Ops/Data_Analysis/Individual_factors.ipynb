```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
nifty = pd.read_csv('/Users/surisettivamsikrishna/Downloads/Vamsi Pc/CODES/Mark 3/Cleaned_Data_UD/Nifty.csv')
nifty['Date'] = pd.to_datetime(nifty['Date'])
```

```python
fdi=pd.read_csv("/Users/surisettivamsikrishna/Downloads/Vamsi Pc/CODES/Mark 3/Cleaned_Data_UD/FDI.csv")
fdi.head(5)
```

```python
# ---> Remove commas and convert to numeric (float)
cols_to_convert = ['Gross Purchase FII', 'Gross Sales FII', 'Net Purchase FII',
                   'Gross Purchase DII', 'Gross Sales DII', 'Net Purchase DII']

for col in cols_to_convert:
    fdi[col] = fdi[col].astype(str).str.replace(',', '', regex=True)  # ---> Remove commas
    fdi[col] = pd.to_numeric(fdi[col], errors='coerce')  # ---> Convert to float

# ---> Check data types after conversion
print(fdi.dtypes)
```

```python


# ---> Plotting Net Purchase FII
plt.plot(fdi['Date'].iloc[:50], fdi['Gross Purchase FII'].iloc[:50], 
         label='Net Purchase FII', marker='o', linestyle='--',color='red')

# ---> Plotting Net Purchase DII
plt.plot(fdi['Date'].iloc[:50], fdi['Gross Purchase DII'].iloc[:50], 
         label='Gross Purchase DII', marker='o', linestyle='--',color='green')

plt.plot(nifty['Date'].iloc[:50],nifty['Close'].iloc[:50],
         label='Nifty', marker='o', linestyle='-',color='blue')

plt.ylim(fdi['Gross Purchase FII'].min(), fdi['Net Purchase FII'].max())


# ---> Formatting
plt.xlabel("Date")
plt.ylabel("Net Purchase")
plt.title("Net Purchase Trends (FII vs DII)")
plt.legend()
plt.grid(True)
plt.xticks(rotation=15)  # ---> Rotates x-axis labels for better readability
plt.yticks(range(0,50000,4000))

plt.show()
```
